## Setup


### Generating a Resend API Key

How to Generate a Resend API Key:

1. **Log in:** Go to the [Resend Login Page](https://resend.com/login) and sign in to your account.
2. **Open API Keys:** Navigate to the **API Keys** section in the left sidebar.
3. **Create the Key:** Click the **Create API Key** button.
4. **Set Permissions:**
   - Give the key a clear name (e.g., "Python Classroom Script").
   - Select **Sending access** for permissions. (This ensures your script can only send emails and cannot accidentally alter your account settings).
5. **Copy and Save:** Copy the generated key (it will start with `re_`).

> **Important:** Resend only shows this key once. Paste it into your notebook secrets immediately. If you close the window before copying it, you must delete that key and create a new one.

Before proceeding, set the API Key as a notebook secret called `RESEND_API_KEY`.

## Sending Email


```python
recipient_address = input("Please enter your email address: ") or "prof.mj.rossetti@gmail.com"
recipient_address
```

```python
import resend

resend.api_key = RESEND_API_KEY

subject = "Hello World"
html = "<p>Congrats on sending your <strong>first email</strong>!</p>"

response = resend.Emails.send({
  "from": SENDER_ADDRESS,
  "to": recipient_address,
  "subject": subject,
  "html": html
})
response
```
